```python
#generate random 2D data using numpy
import numpy as np

data = np.random.randint(1, 100, size=(5, 4))

print(data)
```

```python
# Use pandas features for data cleaning (Hint-RegEx)
import pandas as pd
import numpy as np
import re

# Create DataFrame
df = pd.DataFrame(data)

# 1. Remove extra spaces from Name
df['Name'] = df['Name'].str.replace(r'\s+', ' ', regex=True).str.strip()

# 2. Remove unwanted special characters from Name
df['Name'] = df['Name'].str.replace(r'[^A-Za-z\s]', '', regex=True)

# 3. Convert names to proper/title case
df['Name'] = df['Name'].str.title()

# 4. Clean City names
df['City'] = df['City'].replace({
    'Bombay': 'Mumbai',
    'New Delhi': 'Delhi',
    'NDL': 'Delhi',
    'Chenai': 'Chennai',
    'Chennaai': 'Chennai'
})

# 5. Clean State names
df['State'] = df['State'].replace({
    'Maharastra': 'Maharashtra',
    'Tamilnadu': 'Tamil Nadu'
})

# 6. Fill missing Age with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# 7. Fill missing Ticket Price with median
df['Ticket_Price'] = df['Ticket_Price'].fillna(
    df['Ticket_Price'].median()
)

# 8. Fill missing names
df['Name'] = df['Name'].fillna('Unknown')

# 9. Fill missing Check-In Time
df['Check_In_Time'] = df['Check_In_Time'].fillna('Unknown')

# Display cleaned data
print(df)
```

#Assignment: Visualization for Kumbh Mela Data

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Days and visitor data
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5']

visitors = np.random.randint(
    50000, 200000, size=5
)

# Cities and participant data
cities = ['Prayagraj', 'Varanasi', 'Lucknow', 'Kanpur', 'Agra']

participants = np.random.randint(
    3000, 15000, size=5
)

# Create DataFrames
df_visitors = pd.DataFrame({
    'Day': days,
    'Visitors': visitors
})

df_participants = pd.DataFrame({
    'City': cities,
    'Participants': participants
})

print("Visitors Data:")
display(df_visitors)

print("Participants Data:")
display(df_participants)
```

```python
#1. Line Plot — Number of Visitors Over Different Days
plt.figure(figsize=(8, 5))

plt.plot(
    df_visitors['Day'],
    df_visitors['Visitors'],
    marker='o',
    linestyle='-',
    label='Daily Visitors'
)

plt.xlabel('Days')
plt.ylabel('Number of Visitors')
plt.title('Number of Visitors During Kumbh Mela')

# Gridline
plt.grid(True)

# Legend
plt.legend()

# Save plot
plt.savefig('kumbh_mela_visitors.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
#2. Bar Chart — Number of Participants by City

plt.figure(figsize=(8, 5))

plt.bar(
    df_participants['City'],
    df_participants['Participants'],
    label='Participants'
)

plt.xlabel('Cities')
plt.ylabel('Number of Participants')
plt.title('City-wise Participants in Kumbh Mela')

# Gridlines
plt.grid(axis='y')

# Legend
plt.legend()

# Save plot
plt.savefig('kumbh_mela_city_participants.png',
            dpi=300,
            bbox_inches='tight')

plt.show()
```

```python
#combine both visualizations into one figure
plt.figure(figsize=(14, 5))

# Line Plot
plt.subplot(1, 2, 1)

plt.plot(
    df_visitors['Day'],
    df_visitors['Visitors'],
    marker='o',
    label='Visitors'
)

plt.xlabel('Days')
plt.ylabel('Visitors')
plt.title('Daily Visitors')
plt.grid(True)
plt.legend()

# Bar Chart
plt.subplot(1, 2, 2)

plt.bar(
    df_participants['City'],
    df_participants['Participants'],
    label='Participants'
)

plt.xlabel('Cities')
plt.ylabel('Participants')
plt.title('City-wise Participants')
plt.grid(axis='y')
plt.legend()

plt.tight_layout()

# Save combined visualization
plt.savefig(
    'kumbh_mela_visualization.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
