```python
#Create a list of random numbers and print all the even numbers in the list

import random

# Create a list of 10 random numbers
numbers = [random.randint(1, 100) for _ in range(10)]

print("List:", numbers)

# Print even numbers
print("Even numbers:")

for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
#Create a string and print it in reverse order. Now reverse the words also.

# Create a string
text = "today we are learning python programming in a fun way"

# Reverse the entire string
print("Reverse of string:", text[::-1])

# Reverse the order of words
words = text.split()
print("Reverse of words:", " ".join(words[::-1]))
```
