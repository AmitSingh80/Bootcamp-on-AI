```python
#generate random 2D data using numpy
import numpy as np

data = np.random.randint(1, 100, size=(5, 4))

print(data)
```

```python
# Use pandas features for data cleaning (Hint-RegEx)
import pandas as pd
import numpy as np
import re

# Create DataFrame
df = pd.DataFrame(data)

# 1. Remove extra spaces from Name
df['Name'] = df['Name'].str.replace(r'\s+', ' ', regex=True).str.strip()

# 2. Remove unwanted special characters from Name
df['Name'] = df['Name'].str.replace(r'[^A-Za-z\s]', '', regex=True)

# 3. Convert names to proper/title case
df['Name'] = df['Name'].str.title()

# 4. Clean City names
df['City'] = df['City'].replace({
    'Bombay': 'Mumbai',
    'New Delhi': 'Delhi',
    'NDL': 'Delhi',
    'Chenai': 'Chennai',
    'Chennaai': 'Chennai'
})

# 5. Clean State names
df['State'] = df['State'].replace({
    'Maharastra': 'Maharashtra',
    'Tamilnadu': 'Tamil Nadu'
})

# 6. Fill missing Age with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# 7. Fill missing Ticket Price with median
df['Ticket_Price'] = df['Ticket_Price'].fillna(
    df['Ticket_Price'].median()
)

# 8. Fill missing names
df['Name'] = df['Name'].fillna('Unknown')

# 9. Fill missing Check-In Time
df['Check_In_Time'] = df['Check_In_Time'].fillna('Unknown')

# Display cleaned data
print(df)
```

```python
#use matplotlib to plot a sine wave
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.linspace(0, 2 * np.pi, 100)

# Calculate sine values
y = np.sin(x)

# Plot sine wave
plt.plot(x, y)

# Add labels and title
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.title("Sine Wave")

# Show grid
plt.grid(True)

# Display graph
plt.show()
```
