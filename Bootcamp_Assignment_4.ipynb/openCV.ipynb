```python
import cv2

# 1. Read the image
image = cv2.imread("images.jpeg")

if image is None:
    print("Error: Image not found. Check the image path.")
    exit()

# 2. Resize image
resized = cv2.resize(image, (600, 400))

# 3. Convert to grayscale
gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)

# 4. Crop image
cropped = resized[50:350, 100:500]

# 5. Adjust brightness and contrast
bright = cv2.convertScaleAbs(resized, alpha=1.2, beta=30)

# 6. Apply Gaussian Blur
blur = cv2.GaussianBlur(resized, (7, 7), 0)

# 7. Detect edges
edges = cv2.Canny(resized, 100, 200)

# 8. Draw a rectangle
edited = resized.copy()
cv2.rectangle(
    edited,
    (50, 50),
    (550, 350),
    (0, 255, 0),
    3
)

# 9. Draw a circle
cv2.circle(
    edited,
    (300, 200),
    80,
    (255, 0, 0),
    3
)

# 10. Add text
cv2.putText(
    edited,
    "OpenCV Image Editing",
    (100, 390),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 0, 255),
    2
)

# 11. Save edited images
cv2.imwrite("01_resized.jpg", resized)
cv2.imwrite("02_grayscale.jpg", gray)
cv2.imwrite("03_cropped.jpg", cropped)
cv2.imwrite("04_brightness.jpg", bright)
cv2.imwrite("05_blur.jpg", blur)
cv2.imwrite("06_edges.jpg", edges)
cv2.imwrite("07_final_edited.jpg", edited)

# 12. Display images
cv2.imshow("Original/Resized", resized)
cv2.imshow("Grayscale", gray)
cv2.imshow("Brightness Adjusted", bright)
cv2.imshow("Blurred", blur)
cv2.imshow("Edges", edges)
cv2.imshow("Final Edited Image", edited)

print("Image editing completed successfully!")
print("Edited images have been saved as JPG files.")

cv2.waitKey(0)
cv2.destroyAllWindows()
```
