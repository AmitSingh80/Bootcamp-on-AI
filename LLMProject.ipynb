```python
import random
from collections import defaultdict

# 1. Training Data
text = "AI gives machines the power to see."
words = text.split()

print("Words:", words)

# 2. Build the model
model = defaultdict(list)

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    model[current_word].append(next_word)

print("Model:", dict(model))

# 3. Text Generation
current = "the"
output = [current]

for _ in range(5):
    possible_next_words = model.get(current, [])

    # Stop if there is no next word
    if not possible_next_words:
        break

    next_word = random.choice(possible_next_words)
    output.append(next_word)
    current = next_word

print("Generated text:", " ".join(output))
```
