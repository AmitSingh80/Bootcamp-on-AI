# Spam Detection 

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Toy Training Dataset
training_sentences = [
    "I really enjoyed this movie",
    "The product quality is excellent",
    "The service was amazing and fast",
    "I did not like this movie",
    "The product quality is very poor",
    "The service was slow and disappointing",
]

labels = ["Positive", "Positive", "Positive",
          "Negative", "Negative", "Negative"]

# 2. Convert text sentences into numerical feature vectors
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(training_sentences)

# 3. Train Naive Bayes Classifier
classifier = MultinomialNB()
classifier.fit(X_train, labels)

print("Sentiment Classification Model Trained Successfully!\n")

# 4. Test with unseen sample messages
test_messages = [
    "The movie was excellent and enjoyable",
    "The service was very poor and slow",
]

X_test = vectorizer.transform(test_messages)
predictions = classifier.predict(X_test)

for msg, pred in zip(test_messages, predictions):
    print(f"Message: '{msg}'")
    print(f"Prediction: [{pred}]\n")
```
